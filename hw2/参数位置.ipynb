## 总结：最常修改的参数TOP10
1. **训练脚本**：`num_epochs`、`batch_size`、`init_lr`、`log_iters`
2. **模型结构**：`size_list`（MLP）、CNN层通道数/卷积核尺寸
3. **优化器**：`mu`（动量系数）
4. **正则化**：`weight_decay_lambda`
5. **学习率调度**：`gamma`、`milestones`

所有参数都按上述位置修改即可，无需改动其他代码逻辑。
