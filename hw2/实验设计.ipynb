## 一、全局统一所有超参数（除cnn_momentum初始学习率为0.01外）
| 参数 | 统一值 | 说明 |
|------|--------|------|
| 随机种子 | `np.random.seed(309)` | 固定不变，保证可复现 |
| 批量大小 | `batch_size=128` | 速度最快的批量大小，同时保证梯度稳定 |
| 训练轮数 | `num_epochs=40` | 足够收敛，比20轮节省25%时间 |
| 初始学习率 | `init_lr=0.1` | 所有优化器统一初始学习率 |
| 日志/验证频率 | `log_iters=50` | 单轮打印2次日志，不影响速度 |
| 数据划分 | 训练50000/验证10000/测试10000 | 固定不变 |

## 二、实验设计
### 第一阶段：MLP所有实验
| 序号 | 实验名称 | 对应训练脚本 | 需要修改的参数 | 保存文件名 |
|------|----------|--------------|----------------|------------|
| 1 | MLP基线 | `test_train_mlp.py` | 无需额外修改 | `mlp_baseline.pickle` |
| 2 | SGD+Momentum | `test_train_momentum.py` | 无需额外修改 | `mlp_momentum.pickle` |
| 3 | SGD+StepLR | 复制`test_train.py`为`test_train_mlp_step.py` | 添加：<br>`scheduler = nn.lr_scheduler.StepLR(optimizer, step_size=2000, gamma=0.5)`<br>`runner = RunnerM(..., scheduler=scheduler)` | `mlp_step.pickle` |
| 4 | SGD+ExponentialLR | 复制`test_train.py`为`test_train_mlp_exp.py` | 添加：<br>`scheduler = nn.lr_scheduler.ExponentialLR(optimizer, gamma=0.9999)`<br>`runner = RunnerM(..., scheduler=scheduler)` | `mlp_exp.pickle` |
| 5 | Momentum+StepLR | 复制`test_train_momentum.py`为`test_train_mlp_momentum_step.py` | 添加：<br>`scheduler = nn.lr_scheduler.StepLR(optimizer, step_size=2000, gamma=0.5)`<br>`runner = RunnerM(..., scheduler=scheduler)` | `mlp_momentum_step.pickle` |
| 6 | 仅早停(patience=5) | 复制上面的文件 | 修改：<br>`runner.train(..., early_stop_patience=5)` | `mlp_earlystop.pickle` |
| 7 | 仅L2(λ=1e-4) | 复制上面的文件 | 修改：<br>`linear_model = nn.models.Model_MLP([784,600,10], 'ReLU', lambda_list=[1e-4, 0])` | `mlp_l2_1e-4.pickle` |
